```python
carpeta = Path("datos/ventas")
archivos = sorted(carpeta.glob('*.csv')) 
dfs=[]
for archivo in archivos:
    df = pd.read_csv(archivo, parse_dates=["fecha"])
    df["mes"] = archivo.stem.replace("ventas_", "")
    dfs.append(df)
consolidado= pd.concat(dfs, ignore_index=False)
```

```python
consolidado
```

```python
print(f"archivos consolidados:{len(dfs)}")
```

```python
for grupo in consolidado.groupby("mes")["total"]:
    print(grupo[0])
    print(grupo[1].sum())
```

```python
for grupo1 in consolidado.groupby("producto")["cantidad"]:
    print(grupo1[0])
    print(grupo1[1].sum().max())
    
```

```python
for grupo2 in consolidado.groupby("mes")["total"]:
    print(grupo2[0])
    print(grupo2[1].sum().max())
```

```python
consolidado["total"].mean()
```

```python
consolidado.to_csv('datos/salida/ventas_2024_consolidado.csv',index=False,encoding='utf-8')
```

```python
with pd.ExcelWriter('datos/salida/reporte_anual_2024.xlsx', engine='openpyxl') as writer:
    consolidado.to_excel(writer, sheet_name='Datos_Completos', index=False)
    archivo1.to_excel(writer, sheet_name='Resumen_Mensual', index=False)
```
